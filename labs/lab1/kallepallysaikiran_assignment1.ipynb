```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
df=load_breast_cancer()
x=df.data
y=df.target
```

```python
from collections import Counter
from sklearn.datasets import make_classification
from imblearn.over_sampling import SMOTE
# sm = SMOTE(random_state=42)
# X_res, y_res = sm.fit_resample(x, y)
# X_res=pd.DataFrame(X_res)
# y_res=pd.DataFrame(y_res)


X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
kf=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
model=LogisticRegression(max_iter=1000)
# cross_val=cross_val_score(model,X_train,y_train,scoring="accuracy",cv=kf)
cross_val=cross_val_score(model,X_train,y_train,scoring="accuracy",cv=kf)
```

```python


plt.plot(range(1, 6), cross_val, marker='o', linestyle="-", color='b', label='Accuracy')
plt.axhline(np.mean(cross_val),  linestyle="--",color='r', label=f'Avg: {np.mean(cross_val):.4f}')
plt.xticks(range(1,6))
plt.xlabel("Fold")
plt.ylabel("Accuracy")
plt.title("5-Fold Cross-Validation Accuracy")
plt.legend()
plt.show()

# clf=LogisticRegression(random_state=0).fit(X_train,y_train)

# val=(cross_val_score(clf, X_train, y_train, cv=5))
```

```python
model.fit(X_train,y_train)

y_pred=model.predict(X_test)

# clf.score(X_test,y_test)
accuracy=accuracy_score(y_test,y_pred)
conf_metrix=confusion_matrix(y_test,y_pred)
class_report=classification_report(y_test,y_pred)

print(f"accuracy with test data is {accuracy:.4f}\n")
print(" confussion matrix is \n",conf_metrix)
print("classification report is\n ", class_report)
```

```python
#3 function to give the determinent of a function

import numpy as np
import matplotlib.pyplot as plt

def derivative(f,x,h=1e-5):
  return (f(x+h)-f(x))/h

def plot_the_derivatives(f,x_range):
  x_values=np.linspace(x_range[0],x_range[-1],400)
  y_values=np.array([f(i) for i in x_values])
  dy_values=np.array([derivative(f,x) for x in x_values])

  plt.subplot(1,2,1)
  plt.plot(x_values,y_values,color='b',label='f(x)')
  plt.xlabel("x")
  plt.ylabel("f(x)")
  plt.title("Function")
  plt.legend()
  plt.show()

  plt.subplot(1,2,2)
  plt.plot(x_values,dy_values,color='r',label="f'(x)")
  plt.xlabel("x")
  plt.ylabel("f'(x)")
  plt.title("Derivative")
  plt.legend()
  plt.show()

def f1(x):
  return x**2+2*x+3
def f2(x):
  return np.sin(x)

plot_the_derivatives(f1,(-10,10))
plot_the_derivatives(f2,(-2*np.pi,2*np.pi))
```

```python
#proper code for the question4
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return 3*x**4-6*x**3+12*x**2-24*x+2

def df(x):
  return (12*(x**3))-(18*(x**2))-24


def Gradient_decent(learning_rate=0.01,epoche=1000,tolerance=1e-6):
  x=np.random.uniform(-5,5)
  cost_history=[]

  for _ in range(epoche):
    grad=df(x)
    new_x=x-learning_rate*grad
    cost_history.append(f(x))

    if abs(new_x-x)<tolerance:
      break
    x=new_x
  return x,f(x),cost_history

final_x,final_cost,cost_history=Gradient_decent()
print(f"the final_x is {final_x:.6f}")
print(f"the final_cost is {final_cost:.6f}")

plt.plot(cost_history,label="cost Function",color='b')
plt.xlabel("Epoches")
plt.ylabel("cost function")
plt.title("Cost vs Epochs")
plt.legend()
plt.show()
```
