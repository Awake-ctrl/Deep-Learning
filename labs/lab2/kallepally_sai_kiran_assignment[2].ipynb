# 1 .Perceptron for Binary Classification
Implement the Perceptron algorithm for binary classification. 

a Generate a linearly separable dataset with 5 features, 2 classes, and 1000 samples
using sklearn’s make classification.

b Train your Perceptron model on this dataset and experiment with different
learning rates (e.g., 0.01, 0.1) and iterations (e.g., 50, 100).

c Plot the decision boundary for the best model on training data and evaluate the
model on test data

```python
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=1000,n_features=5,n_classes=2)
X.shape
y.shape
print(X)
print(y)
```

```python
import numpy as  np
from numpy import random
class Perceptron:
    def __init__(self,input_size):
        self.weights= random.rand(1,input_size)
        # self.bias=random.randint(input_size)
        self.bias=1
        print(self.weights)
    def sigmoid(self,z):
        if z<0:
            return 0
        else:
            return 1
        # return (1/1+np.exp(-z))
    def predict(self,value):
        z=np.dot(value,self.weights[0].T)+self.bias
        return self.sigmoid(z)
    def train(self,X,Y,lr,epoche):
        
        for _ in range(epoche):
            # for i in range(len(X)):
            i=random.randint(len(X))
            current_input=X[i]
            current_input=current_input.reshape(1,5)
            target=Y[i]
            prediction=self.predict(current_input)
            error=target-prediction
            weights_update=lr*error
            
            self.weights+=weights_update
            self.bias+=weights_update
            # weights_update=error*prediction*(1-prediction)*current_input
            # self.weights=weights_update*lr
        
    
            # bias_update=error*prediction*(1-prediction)
            # self.bias=bias_update*lr
        
    def evaluates(self,X,y):
        count=0
        
        for i in range(len(X)):
            prediction=self.predict(X[i])
            if prediction==0:
                c=1

            else:
                c=0
            if y[i]==c:
                count+=1
        value=count/len(X)
        print("the accuracy is "+str(value))
            
    def pri(self):
        print("ks")
        print(self.weights[0])
        

```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(

    X, y, test_size=0.33, random_state=42)
```

```python
a=Perceptron(5)
a.train(X_train,y_train,0.05,50)
a.evaluates(X_test,y_test)
```
