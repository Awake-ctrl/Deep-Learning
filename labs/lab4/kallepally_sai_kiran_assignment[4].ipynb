1. (10)Implement a feedforward neural network (FNN) for a classification task with two
hidden layers from scratch, which includes

a Forward propagation from scratch.

b Function for calculating the cross-entropy loss.

c Batch gradient descent with backpropagation from scratch. (Note: Use sigmoid
as the activation function and hidden units of your choice in hidden layers

```python
import numpy as np
class FNN:
    def __init__(self, input_size, hidden_size1, hidden_size2, output_size):
        self.W1 = np.random.randn(input_size, hidden_size1)
        self.b1 = np.zeros((1, hidden_size1))
        self.W2 = np.random.randn(hidden_size1, hidden_size2)
        self.b2 = np.zeros((1, hidden_size2))
        self.W3 = np.random.randn(hidden_size2, output_size)
        self.b3 = np.zeros((1, output_size))
    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))
    def forward(self, X):
        self.a1 = np.dot(X, self.W1) + self.b1
        self.h1 = self.sigmoid(self.a1)
        self.a2 = np.dot(self.a1, self.W2) + self.b2
        self.h2 = self.sigmoid(self.a2)
        self.a3 = np.dot(self.a2, self.W3) + self.b3
        self.h3 = self.sigmoid(self.a3)

        return self.h3


    def cross_entropy_loss(self, y_true, y_pred):
        #three classes
        loss = -np.sum(y_true * np.log(y_pred))

        return loss
    def encoder(self,y):
        y_encoded = np.zeros((y.shape[0], 3))
        y_encoded[np.arange(y.shape[0]), y] = 1
        return y_encoded


    def cross_entropy_derivative(self, y_true, y_pred):
        return y_pred - y_true

    def backward(self, X, y_true, y_pred, learning_rate):
        # m = X.shape[0]

        dz3 = (self.cross_entropy_derivative(y_true=y_true,y_pred=y_pred))*y_pred*(1-y_pred)
        dW3 = np.dot(self.h2.T, dz3)
        db3 = np.sum(dz3, axis=0, keepdims=True)

        dz2 = np.dot(dz3, self.W3.T) * self.h2 * (1 - self.h2)
        dW2 = np.dot(self.h1.T, dz2)
        db2 = np.sum(dz2, axis=0, keepdims=True)

        dz1 = np.dot(dz2, self.W2.T) * self.h1 * (1 - self.h1)
        dW1 = np.dot(X.T, dz1)
        db1 = np.sum(dz1, axis=0, keepdims=True)
        self.W1 -= learning_rate * dW1
        self.b1 -= learning_rate * db1
        self.W2 -= learning_rate * dW2
        self.b2 -= learning_rate * db2
        self.W3 -= learning_rate * dW3
        self.b3 -= learning_rate * db3
        loss = (self.cross_entropy_loss(y_true, y_pred))
        return loss
    def train(self, X, y_true, epochs, learning_rate, batch_size):
        for epoch in range(epochs):
            for i in range(0, X.shape[0], batch_size):
                X_batch = X[i:i+batch_size]
                y_true_batch = y_true[i:i+batch_size]

                y_true_batch=self.encoder(y_true_batch)

                y_pred = self.forward(X_batch)

                # print(y_true_batch)
                # y_pred=np.argmax(y_pred,axis=1)
                # print(y_pred)
                # print(y_true_batch.shape)
                # print(y_pred.shape)
                loss = self.backward(X_batch, y_true_batch, y_pred, learning_rate)
                print(f"Epoch {epoch+1}/{epochs}, Batch {i//batch_size+1}/{X.shape[0]//batch_size}, Loss: {loss:.4f}")
    def predict(self, X):
        y_pred = self.forward(X)
        return y_pred
    def accuracy(self, y_true, y_pred):
        y_pred_class = np.round(y_pred)
        accuracy = np.mean(y_pred_class == y_true)
        return accuracy

```
