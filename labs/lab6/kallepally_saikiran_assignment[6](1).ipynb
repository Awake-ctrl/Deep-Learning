```python
import pandas as pd
```

```python
df=pd.read_csv("/content/data.csv")
Text=df[["Text"]]
Label=df[["Label"]]
```

```python
import nltk
from nltk.corpus import stopwords
import spacy
import string



# Create translation table

translator = str.maketrans('', '', string.punctuation)

# Remove punctuation


nltk.download('stopwords')
words=(stopwords.words('english'))
nlp = spacy.load("en_core_web_sm")
# print(words)
X=[]
x=df["Text"]
for text in x:
  text.lower()
  text= text.translate(translator)

  doc = nlp(text)


  # Remove stopwords
  filtered_words = [token.text for token in doc if not token.is_stop]

  # Join the filtered words to form a clean text
  text = ' '.join(filtered_words)
  X.append(text)


```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenize=Tokenizer()
tokenize.fit_on_texts(X)
sequences = tokenize.word_index

```

```python
print(X[99])
```

```python

y=df["Label"]
input=[]
for i in X:
  input.append(i.split())
```

```python
print(input[0])
```

```python

for i in range(len(input)):

  for j in range(len(input[i])):
    word=input[i][j]
    input[i][j]=sequences[word]



```

```python
import tensorflow.keras
input=keras.utils.pad_sequences(input, padding='post')

from sklearn import preprocessing
lab=preprocessing.LabelEncoder()
df["Label"]=lab.fit_transform(df["Label"])
```

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense,LSTM,Embedding
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow import keras
from sklearn.model_selection import train_test_split
X=input
y=df["Label"]
print(X)

# using the train test split function
X_train, X_test, y_train, y_test = train_test_split(X,y ,
                                   random_state=104,
                                   test_size=0.25,
                                   shuffle=True)




rnn_model = Sequential()
rnn_model.add(Embedding(len(input[0]),100,input_length=len(input[0])))
rnn_model.add(SimpleRNN(50))
rnn_model.add(Dense(6, activation='sigmoid'))

rnn_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])


print(y)
rnn_model.fit(X_train, y_train, epochs=5)
```

```python



# define model
model_lstm = Sequential()
model_lstm.add(Embedding(len(input[0]),100,input_length=len(input[0])))
model_lstm.add(LSTM(64))
model_lstm.add(Dense(6))
model_lstm.compile(optimizer='adam', loss='sparse_categorical_crossentropy',metrics=['accuracy'])
model_lstm.summary()
model_lstm.fit(X_train,y_train, epochs=5)
```
